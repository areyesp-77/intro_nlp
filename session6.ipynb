```python
cadena = 'El corpus paralelo español-náhuatl recopila fuentes de gran variedad, por ejemplo, textos históricos, didácticos, cuentos, recetarios, musicales, entre otros. Los textos que forman el corpus presentan variación dialectal y diacrónica, es decir, no todos los libros pertenecen a la misma variante de náhuatl ni fueron escritos en el mismo periodo.'
```

```python
#lowercase

cadena = cadena.lower()
```

```python
print (cadena)
```

```python
#Simple tokenizer: STR to LIST

cadena = cadena.split()
```

```python
print(cadena) #como se trabaja sobre la misma variable los datos en cadena ahora son LIST
```

```python
#Control structures + Methods

for i in cadena:
  print (i.capitalize())
```

```python
#focus on patterns

for i in cadena:
  if i.startswith('es'):
    print (i)
```

```python
#sufix patterns

for i in cadena:
  if i.endswith('es'):
    print (i.upper(), 'marca sufijal 1')
  elif i.endswith('s'):
    print (i.upper(), 'marca sufijal 2')
  else:
    pass
```

```python
for i in cadena:
  if i.endswith('ble'):
    print (i.upper(), 'flexión adjetival')
  else:
    print ('no hay sufijo adjetival -ble')
```

```python
for i in cadena:
  print (len(i))
```

```python
#LIST to STR
cadena_original = ' '.join(cadena)
print (cadena_original)
```

```python
#Use control structures + method LEN to print the tokens whose length is > N characters

for i in cadena:
  if len(i) >= 8:
    print (i, 'longitud =', len(i))
```

```python
# Frequencies: method COUNT

for i in cadena:
  print (i, 'frecuencia =', cadena.count(i))  #el token se imprime tanta veces sea su frecuencia
```

```python
#Example

key1 = cadena.count('el')
key2 = cadena.count('corpus')
print (key1, key2)
```

```python
#All matters: punctuation marks

key3 = cadena.count('periodo')
print (key3)
```

```python
# Showing the previous statement

frequency_w = 0

for i in cadena:
  frequency_w = cadena.count(i)
  if frequency_w >= 3:
    print (i, 'frecuencia =', frequency_w)

```

```python
# Getting the types + frequency = tuples

frequency_types = {}
for i in cadena:
  frequency_types[i] = cadena.count(i)
print (frequency_types)

```

```python
for i in frequency_types:
  print (i, frequency_types[i])
```

```python
#Getting types whose frequency is > N

for type, frequency in frequency_types.items():
  if frequency >= 2:
    print (type, frequency)
```

```python
# Types in alphabetic order

for i in sorted(frequency_types):
  print (i, frequency_types[i])
```

```python
#Removing punctuation marks 1

signosP = '.,;:[]()$%'
cadena_clean = (''.join(punct for punct in cadena_original if punct not in signosP))
print(cadena_clean)

#notar que este código funciona con el STR original, no con la LIST, por lo tanto, la eliminación de los signos debería hacer antes del método SPLIT
```

```python
#Removing punctuation marks 2 + modules

import string
from string import punctuation

out = str.maketrans('', '', string.punctuation)

cadena_limpia = cadena_original.translate(out)

print (cadena_limpia)
```

```python
cadena = cadena_limpia.split()
print (cadena)
```

```python
frequency_types = {}
for i in cadena:
  frequency_types[i] = cadena.count(i)
print (frequency_types)
```

```python
for type, frequency in frequency_types.items():
  if frequency >= 2:
    print (type, frequency)
```

```python
with open('Murders_Poe.txt', encoding='utf-8') as file:
  corpus = file.read()
  #print(corpus)
```

```python
signosP = '.=,;:"[]()—$%——”“‘'
corpus_clean = (''.join(punct for punct in corpus if punct not in signosP))
print (corpus_clean)
```

```python
corpus_clean = corpus_clean.lower().split()
print (corpus_clean)
```

```python
#Size of corpus
print(len(corpus_clean))
```

```python
#Another way to get the corpus size

size = 0
for i in corpus_clean:
  size = size + 1
print (size)
```

```python
#for i in corpus_clean:
#  if i.startswith('hou'):
#    print(i, '1')
#  elif i.startswith('mur'):
#    print (i, '2')
```

```python
key4 = corpus_clean.count("murder")
print (key4)
```

```python
types_poe = {}
for i in corpus_clean:
  types_poe[i] = corpus_clean.count(i)
print (types_poe)
```

```python
for i in sorted(types_poe):
  print (i, types_poe[i])
```

```python
with open('types.txt', 'w', encoding='utf-8') as file:
  for i in sorted(types_poe):
    file.write(str(i + '\t' + str(types_poe[i]) + '\n'))
```

```python
for i in sorted(types_poe):
  if types_poe[i] >= 100:
    print (i, types_poe[i])
```

```python
#Getting the relative frequency
for i in sorted(types_poe):
  x = types_poe[i]
  y = x/size
  print (f'{y:.8f}')
```

```python
with open('frequencies.txt', 'w', encoding='utf-8') as file:
  for i in sorted(types_poe):
    x = types_poe[i]  # frecuencia absoluta para cada type
    y = x / size      # frecuencia relativa para cada type
    file.write(f'{i}\t{types_poe[i]}\t{y:.8f}\n')
```
