# Métricas de evaluación

Task:

Evaluar el rendimiento de tres POS-taggers respecto del *gold* *standard*.

Usar métricas estándar en PLN para demostrar qué tagger tiene un rendimiento más cercano al juicio de los expertos.

Para estos ejercicios se hará uso de la librería [sckit-learn](https://scikit-learn.org/stable/).



```python
# Importar librerías
import numpy as np
import sklearn
import pandas as pd
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Cargamos el archivo con los datos

df = pd.read_csv('sentiment.csv')
print(df.head(8))
```

## Modelo 1

```python
# Convertir los datos en LIST

goldstd = df.Gold.tolist()
model1 = df.Model1.tolist()

# Accuracy
accuracy_model1 = accuracy_score(goldstd, model1)
print(f"\n--- Accuracy Model 1 ---")
print(f"{accuracy_model1 * 100:.2f}%\n")

# Precision, Recall & F1-Score
print("--- Results ---")
print(classification_report(goldstd, model1, zero_division=0))

# Matriz de confusión
labels = sorted(list(set(goldstd + model1)))
cm = confusion_matrix(goldstd, model1, labels=labels)

# Gráfico
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='inferno', xticklabels=labels, yticklabels=labels)
plt.title('Matriz de confusión')
plt.xlabel('Model 1')
plt.ylabel('Gold Standard')
plt.tight_layout()
plt.show()
```

## Modelo 2

```python
# Convertir los datos en LIST

model2 = df.Model2.tolist()

# Accuracy
accuracy_model2 = accuracy_score(goldstd, model2)
print(f"\n--- Accuracy Model 2 ---")
print(f"{accuracy_model2 * 100:.2f}%\n")

# Precision, Recall & F1-Score
print("--- Results ---")
print(classification_report(goldstd, model2, zero_division=0))

# Matriz de confusión
labels = sorted(list(set(goldstd + model2)))
cm = confusion_matrix(goldstd, model2, labels=labels)

# Gráfico
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='viridis', xticklabels=labels, yticklabels=labels)
plt.title('Matriz de confusión')
plt.xlabel('Model 2')
plt.ylabel('Gold Standard')
plt.tight_layout()
plt.show()
```

## Exploración de resultados mediante DFs y gráficos

```python

# Generar el reporte de clasificación para Modelo 1
model1_dict = classification_report(goldstd, model1, output_dict=True, zero_division=0)
df_report_model1 = pd.DataFrame(model1_dict).transpose()
df_report_model1 = df_report_model1.round(2) # Redondear para mejor legibilidad
df_report_model1 = df_report_model1.add_prefix('M1_')

# Generar el reporte de clasificación para Modelo 2
model2_dict = classification_report(goldstd, model2, output_dict=True, zero_division=0)
df_report_model2 = pd.DataFrame(model2_dict).transpose()
df_report_model2 = df_report_model2.round(2)
df_report_model2 = df_report_model2.add_prefix('M2_')

# Generar el reporte de clasificación para Modelo 3
model3_dict = classification_report(goldstd, model3, output_dict=True, zero_division=0)
df_report_model3 = pd.DataFrame(model3_dict).transpose()
df_report_model3 = df_report_model3.round(2)
df_report_model3 = df_report_model3.add_prefix('M3_')

# Generar el reporte de clasificación para Modelo 4
model4_dict = classification_report(goldstd, model4, output_dict=True, zero_division=0)
df_report_model4 = pd.DataFrame(model4_dict).transpose()
df_report_model4 = df_report_model4.round(2) # Redondear para mejor legibilidad
df_report_model4 = df_report_model4.add_prefix('M4_')

# Unir los DataFrames para una comparación de los tres taggers
df_comparison = pd.concat([df_report_model1, df_report_model2, df_report_model3, df_report_model4], axis=1)

```

```python
from google.colab import files

# Graficar valores de Precision

# Filtrar las filas de comparison (accuracy, macro avg, weighted avg)
df_scores = df_comparison.drop(index=['macro avg', 'weighted avg', 'accuracy'])

# Seleccionar solo las columnas de Precision
df_precision = df_scores[['M1_precision', 'M2_precision', 'M3_precision', 'M4_precision']]

# Reorganizar el DF
df_precision_combined = df_precision.reset_index().melt(id_vars='index', var_name='Model', value_name='Precision')
df_precision_combined['Model'] = df_precision_combined['Model'].str.replace('_precision', '')

# Plot
plt.figure(figsize=(9, 6))
sns.barplot(x='index', y='Precision', hue='Model', data=df_precision_combined, palette='YlGnBu')#bone #PuBu #ocean
plt.title('Comparación de valores de Precision')
plt.xlabel('Models')
plt.ylabel('Precision')
plt.xticks(rotation=45, ha='right') # Rotar etiquetas para mejor legibilidad
plt.ylim(0, 1)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='Precision', loc='best', bbox_to_anchor=(1, 1) )
plt.tight_layout()
plt.savefig('precision.png')
plt.show()

files.download('precision.png')
```

```python
# Graficar valores de Recall

# Seleccionar solo las columnas de Recall
df_recall = df_scores[['M1_recall', 'M2_recall', 'M3_recall', 'M4_recall']]

# Reorganizar el DF
df_recall_combined = df_recall.reset_index().melt(id_vars='index', var_name='Model', value_name='Precision')
df_recall_combined['Model'] = df_recall_combined['Model'].str.replace('_recall', '')

# Plot
plt.figure(figsize=(9, 6))
sns.barplot(x='index', y='Precision', hue='Model', data=df_precision_combined, palette='YlGnBu')#bone #PuBu #ocean
plt.title('Comparación de valores de Recall')
plt.xlabel('Models')
plt.ylabel('Recall')
plt.xticks(rotation=45, ha='right') # Rotar etiquetas para mejor legibilidad
plt.ylim(0, 1)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='Recall', loc='best', bbox_to_anchor=(1, 1) )
plt.tight_layout()
plt.savefig('recall.png')
plt.show()

files.download('recall.png')
```

```python
# Graficar valores de F1

# Seleccionar solo las columnas de F1
df_f1 = df_scores[['M1_f1-score', 'M2_f1-score', 'M3_f1-score', 'M4_f1-score']]

# Reorganizar el DF
df_f1_combined = df_f1.reset_index().melt(id_vars='index', var_name='Model', value_name='F1-score')
df_f1_combined['Model'] = df_f1_combined['Model'].str.replace('_f1-score', '')

# Plot
plt.figure(figsize=(9, 6))
sns.barplot(x='index', y='F1-score', hue='Model', data=df_f1_combined, palette='YlGnBu')#bone #PuBu #ocean
plt.title('Comparación de valores de F1')
plt.xlabel('Models')
plt.ylabel('F1')
plt.xticks(rotation=45, ha='right') # Rotar etiquetas para mejor legibilidad
plt.ylim(0, 1)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='F1', loc='best', bbox_to_anchor=(1, 1) )
plt.tight_layout()
plt.savefig('f1.png')
plt.show()

files.download('f1.png')
```

```python
# Comparación de rendimiento de los mejores taggers

# Selección de datos = F1
df_f1_scatter = df_f1[['M1_f1-score', 'M3_f1-score']].copy()
plt.figure(figsize=(10, 8))
sns.scatterplot(x='M1_f1-score', y='M3_f1-score', data=df_f1_scatter, s=100, hue=df_f1.index, palette='viridis')

# Añadir etiquetas a cada punto
for i, row in df_f1_scatter.iterrows():
    plt.text(row['M1_f1-score'] + 0.01, row['M3_f1-score'] + 0.01, i, fontsize=9)

# Añadir una línea diagonal para la referencia (y=x)
max_f1 = df_f1_scatter.max().max()
plt.plot([0, max_f1], [0, max_f1], color='red', linestyle='dotted', label='Mismo comportamiento')

plt.title('Comparación de F1 entre M1 y M3')
plt.xlabel('F1 de M1')
plt.ylabel('F1 de M3')
plt.xlim(-0.05, max_f1 + 0.1) # Ajustar límites para que las etiquetas sean visibles
plt.ylim(-0.05, max_f1 + 0.1)
plt.grid(True, linestyle='--')
plt.legend(title='Best models', loc='best', bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.savefig('rendimientoF1.png')
plt.show()


files.download('rendimientoF1.png')
```
