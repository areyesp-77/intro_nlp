```python
#Open several files and save their content in independent variables: variable1 to variableN

# Dictionary to store file contents in independent variables
file_contents_dict = {}

# Iterate through the files found by glob.glob
for input_file_path in files:
    # Extract the filename from the path
    filename = os.path.basename(input_file_path)

    try:
        # Read the content of the input file
        with open(input_file_path, 'r', encoding='utf-8') as f_in:
            content = f_in.read()

        # Store the content in the dictionary with the filename as the key
        file_contents_dict[filename] = content

        # Construct the dictionary access string for clearer output
        dict_access_str = f"file_contents_dict['{filename}']"
        print(f"Content of '{filename}' loaded into {dict_access_str}")
    except Exception as e:
        print(f"Error processing file '{filename}': {e}")

# You can now access the content of each file like: file_contents_dict['file00001.txt']
```

```python
import string
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# You have to download the set of punctuation marks and stopwords first.
# This process must be carried out just once, then comment the lines.
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')

# Lowercasing and tokenizing independent variables
for i, j in file_contents_dict.items():
  files_content = j.lower() #in order to use the NLTK method, we apply lowercasing first over the content (j)
  tokens = word_tokenize(files_content) #then, we tokenize using the NLTK method


  # Set the punctuation marks and stopwords in variables
  stop_words = stopwords.words('english')  #we use the NLTK method
  punctuation = set(string.punctuation)    # we use the STRING method

  # Removing punctuation marks and stopwords
  tokens_clean = [word for word in tokens if word.lower() not in punctuation and word.lower() not in stop_words]
  print(f"File ID: {i}")
  print(f"Size: {len(tokens_clean)}")

  # Types and frequency
  types_freq = {}
  for word in tokens_clean:
      types_freq[word] = types_freq.get(word, 0) + 1

  print(f"Types & freq): {sorted(types_freq.items(), key=lambda item: item[1], reverse=True)}")

  print("\n" + "-"*50 + "\n") #format to clarify view

with open('../output.txt', 'w', encoding='utf-8') as file:
      for word, frequency in sorted(types_freq.items(), key=lambda item: item[1], reverse=True):
          file.write(f"{word}\t{frequency}\n")
```

```python
import string
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

#List 2 string
cadena = ' '.join(files)
#print (cadena)

# Tokenizing corpus
words = word_tokenize(cadena.lower()) #we use a NLTK method
print(words)
print(len(words))

# Set the punctuation marks and stopwords in variables
stop_list = stopwords.words('english')  #we use the NLTK method
p_marks = set(string.punctuation)    # we use the STRING method

  # Removing punctuation marks and stopwords
words_clean = [word for word in words if word.lower() not in p_marks and word.lower() not in stop_list]
print(words_clean)
print(len(words_clean))

# Types and frequency
words_freq = {}
for word in words_clean:
  words_freq[word] = words_freq.get(word, 0) + 1
print(sorted(words_freq.items(), key=lambda item: item[1], reverse=True))


# Types and frequency as we learned
word_types = {}
for word in words_clean:
  word_types[word] = words_clean.count(word)
print(sorted(words_freq.items(), key=lambda item: item[1], reverse=True))
```
