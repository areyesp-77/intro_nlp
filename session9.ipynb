## Gold Standard

De acuerdo con Jurafsky & Martin (2026), el gold standard es una "human-labeled reference"; i.e., la formalización del conocimiento por medio del juicio de un humano. Por ejemplo, el humano debe ser capaz de identificar por qué X es X y qué es lo que la diferencia de Y. Si lo vemos en términos lingüísticos, por qué *bajo* es un *verbo* y no un *adjetivo*.

Tener un gold standard no siempre es posible. Hay tareas en las cuales resulta muy complicado/costoso tener un dataset que represente fielmente el fenómeno de estudio, por ejemplo, ironía o *grooming*.

Para el caso de la actividad, nosotros generaremos un gold standard que servirá para evaluar la calidad de anotación de tres POS taggers.

# Inter-annotator agreement

¿Qué tanta coincidencia existe entre las etiquetas asignadas por los humanos?

Hay diversas métricas, por ejemplo:
* Porcentaje de acuerdo (múltiples anotadores, acuerdo por casualidad, baseline)

* [Kappa de Cohen](https://real-statistics.com/reliability/interrater-reliability/cohens-kappa/) (entre dos anotadores)
* [Kappa de Fleiss](https://real--statistics-com.translate.goog/reliability/interrater-reliability/fleiss-kappa/?_x_tr_sl=en&_x_tr_tl=es&_x_tr_hl=es&_x_tr_pto=tc) (entre tres o más anotadores - datos categoriales)
* [Alfa de Krippendorff](https://real-statistics.com/reliability/interrater-reliability/krippendorffs-alpha/krippendorffs-alpha-basic-concepts/) (múltiples anotadores, etiquetas faltantes, diversos tipos de datos)

# Ejecución de POS taggers

````python
import pandas as pd

with open('data.txt', encoding='utf-8') as f:
  corpus = f.read()
  print (corpus)
````

````python
# TreeTagger
!cat data.txt | cmd/tree-tagger-spanish > treetagger.tsv
````

````python
# spaCy

df_spacy = pd.DataFrame(columns=['Token', 'POS'])

model1 = nlp(corpus)
data_spacy = []
for i in model1:
  data_spacy.append({'Token': i.text, 'POS': i.tag_})
df_spacy = pd.DataFrame(data_spacy)
print(df_spacy)
df_spacy.to_csv('spacy.csv', index=False)
````

````python
# Stanza
df_stanza = pd.DataFrame(columns=['Token', 'POS'])

model2 = pln(corpus)
data_stanza = []
for i, s in enumerate(model2.sentences):
  for word in s.words:
    data_stanza.append({'Token': word.text, 'TAG': word.upos})
df_stanza = pd.DataFrame(data_stanza)
print(df_stanza)
df_stanza.to_csv('stanza.csv', index=False)
````

````python
df = pd.read_csv('gold_standard.csv')
df1 = pd.read_csv('treetagger.tsv', sep='\t', header=None, usecols=[1])
df2 = pd.read_csv('spacy.csv', usecols=['POS'])
df3 = pd.read_csv('stanza.csv', usecols=['TAG'])
````

````python
df_assessment = pd.concat([df, df1, df2, df3], axis=1)
df_assessment.columns = ['W', 'Gold_STD', 'TreeTger', 'Spacy', 'Stanza']

df_final = df_assessment.replace({
    'ADV':'ad',
    'NOUN':'n',
    'VERB':'v',
    'ADJ':'aj',
    'PRON':'pr',
    'PREP':'p',
    'DET':'d',
    'CONJ':'c',
    'INTJ':'i',
    'X':'uk',
    'ART': 'd',
    'ADP': 'p',
    'PUNCT': 'uk',
    'NC': 'n'
})
````
