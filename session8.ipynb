
#HMM
	Conjunto de estados Q
	Matriz de transición de probabilidad A
	Matriz de emisión de probabilildad B  
    Distribución de probabilidad inicial π

## Fundamento
A diferencia del modelo de Markov, en los HMM hay una capa más: la matriz B (emisión)

###Ejemplo
* Observaciones visibles = la casa es azul
* Estado oculto = {POS, POS, POS, POS} > {D, N, V, AJ}

Los estados ocultos se desconocen; i.e. no se sabe cuál es la posible etiqueta que le correspnde a cada palabra

## Distribución de probabilidad inicial

* $π_d$, $π_n$, $π_v$, $π_{aj}$
* [0.80. 0.05, 0.10, 0.05]

argmax = probabilidad inicial * matriz b * matriz a


$$\arg\max_{t_1, \dots, t_n} \prod_{i=1}^{n} \text{emission}(t_i) \cdot \text{transition}(t_{i-1}, t_i)$$


Es decir: probabilidad inicial * probabilidad de emisión * probabilidad de transición

---> Input = La casa es azul = d n v aj

---> Output = d n v aj



---

I(d) = 0.80

x E (det = la) = 0.95

x E (n = casa) = 0.95

x T (n > v ) = 0.40

x E (v = es) = 0.7

x T (v > aj) = 0.35

x E (aj = azul) = 1

Producto = 0.0672182
